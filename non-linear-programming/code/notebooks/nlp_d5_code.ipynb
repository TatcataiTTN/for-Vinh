## Cặp LP chuẩn và giá bóng (dạng A, B)

Bài lớn 5.1: x*=(2,0,1), y*=(1,0,1), cùng giá trị 13.

```python
import numpy as np
from scipy.optimize import linprog
A = np.array([[2,3,1],[4,1,2],[3,4,2.]]); b = np.array([5,11,8.]); c = np.array([5,4,3.])
p = linprog(-c, A_ub=A, b_ub=b, bounds=[(0,None)]*3, method="highs")
d = linprog(b, A_ub=-A.T, b_ub=-c, bounds=[(0,None)]*3, method="highs")
print("x* =", p.x, " p* =", -p.fun); print("y* =", d.x, " d* =", d.fun)
for t in (4, 5, 6): print("b1 =", t, round(-linprog(-c, A_ub=A, b_ub=[t,11,8], bounds=[(0,None)]*3, method="highs").fun, 4))
```

## Hàm đối ngẫu Bài tập 1 (dạng C)

q(λ)=3λ−3λ²/8, λ*=4, d*=6; gốc (2,1).

## Dual ascent trên Ví dụ 1 (dạng F)

α=0.5 hội tụ về 2; α=2 dao động.

```python
for alpha in (0.5, 1.0, 2.0):
    lam = 0.0; out = []
    for k in range(6):
        x = lam/2; g = 2 - 2*x; out.append(round(lam, 4)); lam = max(0.0, lam + alpha*g)
    print(alpha, out)
```

## Bài code · Chiếu và nhân tử cho min a x² + b y², x+y ≥ c

Đầu vào: $a,b,c>0$. Đầu ra: $\lambda^*,x^*,y^*,p^*$ (dùng công thức đóng).

```python
def solve(a, b, c):
    s = 1/a + 1/b; lam = 2*c/s
    return lam, lam/(2*a), lam/(2*b), c*c/s
print(solve(1, 2, 3))   # Bài tập 1: (4, 2, 1, 6)
print(solve(1, 1, 2))   # Ví dụ 1:   (2, 1, 1, 2)
```

## Bài code · Một bước dual ascent

Đầu vào: $\lambda,\alpha$. Đầu ra: $\lambda$ mới cho Ví dụ 1.

```python
def step(lam, alpha): return max(0.0, lam + alpha*(2 - lam))
print(step(0, 0.5), step(1, 0.5), step(0, 2))   # 1.0, 1.5, 4.0
```
