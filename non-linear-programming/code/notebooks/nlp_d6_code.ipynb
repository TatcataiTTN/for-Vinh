## Ví dụ 4: nghiệm và nhân tử bằng cvxpy (dạng B)

Nghiệm (1,1), λ của x+y≤2 bằng 2.

```python
import numpy as np, cvxpy as cp
x = cp.Variable(2); c = [x[0] + x[1] <= 2, x >= 0]
p = cp.Problem(cp.Minimize(cp.sum_squares(x - 2)), c); p.solve()
print("x* =", x.value.round(4), " f* =", round(float(p.value), 4), " lambda1 =", round(float(c[0].dual_value), 4))
```

## Ví dụ 3: ba điểm KKT bằng SLSQP (dạng B, C)

Từ nhiều điểm đầu thu được (±1,∓1) và (0,0) khi bắt đầu đúng điểm.

```python
import numpy as np
from scipy.optimize import minimize
cons = [{"type": "ineq", "fun": lambda v: 2 - v[0]**2 - v[1]**2}]
pts = {(round(r.x[0], 3), round(r.x[1], 3), round(r.fun, 3)) for r in (minimize(lambda v: v[0]*v[1], np.random.RandomState(s).uniform(-1.4, 1.4, 2), constraints=cons, method="SLSQP") for s in range(20))}
print(sorted(pts))
```

## Kiểm điểm KKT của Ví dụ 2 (dạng B)

Giải nhân tử tại đỉnh (1/3, 8/3): λ2 âm.

```python
import numpy as np
x, y = 1/3, 8/3
gF = np.array([-2*x - 4, -2*y + 6]); G = np.array([[1, 1], [-2, 1.]])
lam = np.linalg.solve(G.T, -gF)
print("lambda =", lam.round(4))
```

## Bài code · Giải bảng ca KKT cho QP hai biến

Đầu vào: tâm $c=(a,b)$, hằng số $s$ cho ràng buộc $x+y\le s$, $x,y\ge0$. Đầu ra: nghiệm và $\lambda$ (dùng hình chiếu).

```python
import numpy as np
def proj(c, s):
    c = np.array(c, float)
    if c.sum() <= s and (c >= 0).all(): return c, 0.0
    t = (c.sum() - s) / 2; x = c - t
    return np.maximum(x, 0), c.sum() - s
print(proj((2, 2), 2))   # Ví dụ 4: (1,1), lambda=2
print(proj((1, 0.5), 3))
```

## Bài code · Phân loại điểm KKT bằng đường đi

Đầu vào: hàm f, điểm p và hướng d. Đầu ra: dấu $f(p+td)-f(p)$ với $t$ nhỏ theo hai hướng.

```python
f = lambda x, y: x*y
t = 1e-3
print(f(t, t) - f(0, 0), f(t, -t) - f(0, 0))   # (+, -) ⇒ yên ngựa
```
